# 01 — Coleta e leitura dos dados RONI

Coleta da série completa do **RONI** publicada pela NOAA/CPC e conversão do TXT para CSV.

**Fonte:** https://www.cpc.ncep.noaa.gov/data/indices/RONI.ascii.txt

Execute as células em ordem, com o diretório de trabalho na pasta `notebooks/`, como no notebook do BDQueimadas. A coleta precisa de internet e atualiza `dados/RONI.ascii.txt` e `dados/roni.csv`.


## 1. Bibliotecas e pasta de dados

Usamos `../dados` para acessar a pasta de dados a partir de `notebooks/`. Se o pandas ainda não estiver instalado no kernel, execute `%pip install pandas` em uma célula antes de continuar.


```python
from pathlib import Path
from urllib.request import urlretrieve

import pandas as pd

DATA_DIR = Path("../dados")
DATA_DIR.mkdir(parents=True, exist_ok=True)

URL_RONI = "https://www.cpc.ncep.noaa.gov/data/indices/RONI.ascii.txt"
ARQUIVO_TXT = DATA_DIR / "RONI.ascii.txt"
ARQUIVO_CSV = DATA_DIR / "roni.csv"
```

## 2. Baixar o TXT


```python
urlretrieve(URL_RONI, ARQUIVO_TXT)
print(f"TXT salvo em: {ARQUIVO_TXT}")
```

## 3. Ler o TXT e salvar como CSV

O separador `r"\s+"` permite ler as colunas separadas por espaços no TXT. Renomeamos `SEAS` para `trimestre`, `YR` para `ano` e `ANOM` para `roni`.

Os trimestres são sobrepostos: `DJF` representa dezembro–janeiro–fevereiro, `JFM` representa janeiro–fevereiro–março, e assim por diante. Os valores correspondem a trimestres, não a meses isolados.


```python
df = pd.read_csv(ARQUIVO_TXT, sep=r"\s+")
df = df.rename(columns={"SEAS": "trimestre", "YR": "ano", "ANOM": "roni"})

df.to_csv(ARQUIVO_CSV, index=False, encoding="utf-8", float_format="%.2f")
print(f"CSV salvo em: {ARQUIVO_CSV} — {len(df)} registros")
```
