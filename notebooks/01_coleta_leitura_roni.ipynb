# 01 - Coleta e leitura do índice RONI

Download da série histórica do índice RONI em formato TXT na web e conversão para `dados/roni.csv`.

## 1. Coleta do arquivo TXT na web

```python
from urllib.request import urlopen
import pandas as pd

url = "https://www.cpc.ncep.noaa.gov/data/indices/RONI.ascii.txt"

with urlopen(url) as resposta:
    texto = resposta.read().decode("utf-8")

with open("../dados/RONI.ascii.txt", "w", encoding="utf-8") as f:
    f.write(texto)

print("Arquivo salvo em ../dados/RONI.ascii.txt")
```

## 2. Conversão para CSV

Cada linha representa um trimestre móvel sobreposto (`DJF` para dezembro-janeiro-fevereiro, `JFM` para janeiro-fevereiro-março e seguintes).

```python
df = pd.read_csv("../dados/RONI.ascii.txt", sep=r"\s+")
df = df.rename(columns={"SEAS": "trimestre", "YR": "ano", "ANOM": "roni"})
df.to_csv("../dados/roni.csv", index=False)

print("Registros salvos em ../dados/roni.csv:", len(df))
```
