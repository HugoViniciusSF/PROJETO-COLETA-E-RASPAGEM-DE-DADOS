# 01 — Coleta e Leitura dos Arquivos do INPE

Leitura inicial e exploração estrutural das duas bases de focos de queimadas do INPE (BDQueimadas, satélite `NOAA-20`):
- **Semiestruturado (GeoJSON):** `dados/bdqueimadas_2024-11-30_2025-11-30.geojson`
- **Estruturado (CSV):** `dados/bdqueimadas_2025-12-01_2026-11-19.csv`

## 1. Bibliotecas e configuração de ambiente (Local / Google Colab)

```python
import importlib.util
from pathlib import Path

import geopandas as gpd
import pandas as pd

IN_COLAB = importlib.util.find_spec("google.colab") is not None

if IN_COLAB:
    repo_dados = Path("/content/PROJETO-COLETA-E-RASPAGEM-DE-DADOS/dados")
    if repo_dados.exists():
        DATA_DIR = repo_dados
    else:
        from google.colab import drive

        drive.mount("/content/drive")
        DATA_DIR = Path("/content/drive/MyDrive/PROJETO-COLETA-E-RASPAGEM-DE-DADOS/dados")
else:
    DATA_DIR = Path("../dados")

ARQUIVO_GEOJSON = DATA_DIR / "bdqueimadas_2024-11-30_2025-11-30.geojson"
ARQUIVO_CSV = DATA_DIR / "bdqueimadas_2025-12-01_2026-11-19.csv"

print(f"Ambiente: {'Google Colab' if IN_COLAB else 'Local'}")
print(f"GeoJSON encontrado: {ARQUIVO_GEOJSON.exists()} ({ARQUIVO_GEOJSON})")
print(f"CSV encontrado:     {ARQUIVO_CSV.exists()} ({ARQUIVO_CSV})")
```

## 2. Leitura do arquivo semiestruturado (GeoJSON)

```python
df_geojson = gpd.read_file(ARQUIVO_GEOJSON)
print("GeoJSON carregado com sucesso.")
```

## 3. Leitura do arquivo estruturado (CSV)

```python
df_csv = pd.read_csv(ARQUIVO_CSV, encoding="utf-8-sig")
print("CSV carregado com sucesso.")
```

## 4. Conferência das colunas entre os dois DataFrames

```python
cols_geojson = list(df_geojson.columns)
cols_csv = list(df_csv.columns)

set_geojson = set(cols_geojson)
set_csv = set(cols_csv)

print("Colunas GeoJSON:", cols_geojson)
print("Colunas CSV:    ", cols_csv)
print("Mesmas colunas (exatas)?", set_geojson == set_csv)
print("Em comum:          ", sorted(set_geojson & set_csv))
print("Somente no GeoJSON:", sorted(set_geojson - set_csv))
print("Somente no CSV:    ", sorted(set_csv - set_geojson))
```
