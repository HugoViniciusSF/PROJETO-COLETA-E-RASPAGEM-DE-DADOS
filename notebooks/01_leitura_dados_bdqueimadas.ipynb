# 01 - Leitura dos dados do BDQueimadas (INPE)

Leitura e conferência estrutural dos dois arquivos de focos de queimadas do satélite `NOAA-20`: o GeoJSON de `2024-11-30` a `2025-11-30` e o CSV de `2025-12-01` a `2026-11-19`.

## 1. Leitura do GeoJSON e do CSV

```python
import geopandas as gpd
import pandas as pd

df_geojson = gpd.read_file("../dados/bdqueimadas_2024-11-30_2025-11-30.geojson")
df_csv = pd.read_csv("../dados/bdqueimadas_2025-12-01_2026-11-19.csv", encoding="utf-8-sig")

print("Dimensões GeoJSON:", df_geojson.shape)
print("Dimensões CSV:    ", df_csv.shape)
```

## 2. Conferência das colunas

```python
print("Colunas GeoJSON:", list(df_geojson.columns))
print("Colunas CSV:    ", list(df_csv.columns))

df_geojson = df_geojson.drop(columns=["geometry"])
print("Mesmas 12 colunas?", list(df_geojson.columns) == list(df_csv.columns))
```

## 3. Estrutura do GeoJSON (2024-2025)

```python
display(df_geojson.head())
print("Tipos de dados:")
display(df_geojson.dtypes)
df_geojson.info()
```

## 4. Estrutura do CSV (2025-2026)

```python
display(df_csv.head())
print("Tipos de dados:")
display(df_csv.dtypes)
df_csv.info()
```
